    **Увага!** Для обчислення стандартного відхилення маємо скористатись функцією `np.std(your_dataframe, ddof=1)`. Чому так - розбираємо в лекції про t-test.

```python
from scipy import stats

# Крок 1: параметри задачі
sample_mean = np.mean(data)
population_mean = 20
population_std = np.std(data, ddof=1)
sample_size = len(data)
alpha = 0.05

print(f"sample_mean = {sample_mean:.4f}")
print(f"population_std = {population_std:.4f}")
print(f"sample_size = {sample_size}")
```

```python
# Крок 2: гіпотези
# H0: mu = 20  (середній чек не змінився)
# Ha: mu > 20  (середній чек підвищився)
# Тип тесту: правосторонній (one-tailed, right)
```

```python
# Крок 3: стандартна помилка SE
SE = population_std / np.sqrt(sample_size)
print(f"SE = {SE:.4f}")
```

```python
# Крок 4: z-статистика
z_stat = (sample_mean - population_mean) / SE
print(f"z-статистика = {z_stat:.4f}")
```

```python
# Крок 5: критичне z-значення (правосторонній тест)
z_critical = stats.norm.ppf(1 - alpha)
print(f"z_critical = {z_critical:.4f}")
```

```python
# Крок 6: p-value (правосторонній тест)
p_value = 1 - stats.norm.cdf(z_stat)
print(f"p-value = {p_value:.4f}")
```

```python
# Крок 7: рішення
print(f"\nz_stat ({z_stat:.4f}) vs z_critical ({z_critical:.4f})")
print(f"p-value ({p_value:.4f}) vs alpha ({alpha})")

if z_stat > z_critical:
    print("Відхиляємо H0: середній чек статистично значущо підвищився")
else:
    print("Не відхиляємо H0: недостатньо доказів для підвищення")
```

```python
from statsmodels.stats.weightstats import ztest

z_stat2, p_value2 = ztest(data, value=population_mean, alternative='larger')

print(f"z-статистика = {z_stat2:.4f}")
print(f"p-value = {p_value2:.4f}")

if p_value2 < alpha:
    print("Відхиляємо H0: середній чек статистично значущо зріс")
else:
    print("Не відхиляємо H0: недостатньо доказів")

print(f"\nЗбіг з Завданням 1: z={z_stat:.4f} vs {z_stat2:.4f}, p={p_value:.4f} vs {p_value2:.4f}")
```
